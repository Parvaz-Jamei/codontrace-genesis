```python
import json
from pathlib import Path
from codontrace.genesis.claim_gate import ClaimRequest, ScientificClaimGate

root = Path('/content/drive/MyDrive/codontrace_genesis/engine_stress_colab')
out = sorted(root.glob('run_*'))[-1]
print('OUT', out)
for name in ['summary.json', 'claim_readiness.json']:
    p = out / name
    print('===', name, p.is_file(), 'bytes', p.stat().st_size if p.is_file() else 0)

gate = ScientificClaimGate()
decisions = {}
for label in [
    'intervention_supported',
    'collective_intelligence_candidate',
    'collective_intelligence',
    'proved_collective_intelligence',
    'agi',
    'tokyo_type1_passed',
]:
    d = gate.decide(ClaimRequest(label, {}))
    decisions[label] = bool(d.allowed)
    print(label, 'allowed=', d.allowed)

(out / 'CLAIM_GATE_EMPTY_FLAGS.json').write_text(json.dumps(decisions, indent=2) + '\n')
(out / 'DONE.txt').write_text('ok\n')
print('Wrote DONE.txt — tell the agent folder:', out)
```
